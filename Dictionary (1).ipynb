```python
1.# What is a Dictionary in Python?

#A dictionary is a collection of key-value pairs
#create a dictionary
student={
         "name":"Ram",
          "age":15,
          "marks":34
}
print(student)
```

```python
#access dictionary
student={
         "name":"rahul",
          "age":45,
          "marks":67
}
print(student.get("age"))
print(student.get("name"))    
print(student["marks"])
```

```python
# Difference Between get() and []
data={"name":"neetu"}
print(data.get("age"))
print(data["age"])
```

```python
#Add New Item
student={"name":"raj"}
student["age"]=22
print(student)
```

```python
#. Update Dictionary
student={"name":"ram","age":20}
student["age"]=22
print(student)
```

```python
#Delete Item from Dictionary
student={"name":"ratan","age":21}
del student["age"]
print(student)
#Using pop():
student.pop("name")
```

```python
#Loop Through Dictionary
student={
    "name":"ratan",
    "age":23,
    "marks":34,
    "address":"delhi"
}
for key,value in student.items():
    print(key,value)
    
```

```python
#Nested Dictionary
students={
          "s1":{"name":"ram","age":23,"marks":56},
          "s2":{"name":"ekta","age":12,"marks":34}
}
print(students["s1"]["name"])


```

```python
# Dictionary Length
data = {"a":1, "b":2, "c":3}
print(len(data))
```

```python
#Check Key Exists or Not
# Copy Dictionary
d1={"name":"neha","age":23}
d2=d1.copy()
d2
```

```python
# Empty Dictionary
data={}
print(type(data))
```

```python
#Dictionary Using dict() Constructor
student = dict(name="Aman", age=20)

print(student)
```

```python
# Most Important Program
#Program 1: Student Information
student = {
    "name":"Rahul",
    "roll":101,
    "marks":89
}

print(student)
```

```python
#Program 2: Sum of Dictionary Values
data = {
    "a":10,
    "b":20,
    "c":30
}

print(sum(data.values()))
```

```python
#Program 3:frequency of character
word = input("Enter a word: ")

freq = {}

for ch in word:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print(freq)

```

```python
#Program 4: Merge Two Dictionaries
d1 = {"a":1}
d2 = {"b":2}

d1.update(d2)

print(d1)
```
